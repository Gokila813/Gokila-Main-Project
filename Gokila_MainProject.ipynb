# Project Title

Diabetes Health Risk Analysis Dataset Clinical and lifestyle factors associated with diabetes status.

# **Objective(Problem statment)**

1. To identify meaningful patterns, trends, and insights from the diabetes dataset using descriptive analysis (EDA)
2. To clean, transform, and preprocess the data for analysis
3. To diagnose the key clinical and lifestyle factors associated with diabetes using diagnostic analysis
4. To create meaningful visualizations that clearly explain the relationship between health/lifestyle factors and diabetes
5. To provide actionable insights and recommendations based on the analysis

# Outcome

The project helps uncover key patterns and relationships between clinical indicators (Glucose, BMI, Insulin) and lifestyle factors (Diet, Exercise, Sleep, Stress) influencing diabetes. The analysis provides a clear understanding of "what is happening" and "why it is happening," supporting evidence-based, health-related decision-making.

#**Dataset Information**

**Source:** UCI ML Repository / Government Portals / Company Internal Dataset / Open Data Portals <font color="blue"><u> **https://www.kaggle.com/datasets/miadul/diabetes-risk-and-lifestyle-factors**</font></u>

**Year / Timeline:** Data collected during YYYY to YYYY

**Dataset Column/features Description:**

**Glucose:** Blood glucose level of the individual.  
**BMI:** Body Mass Index, indicating body weight relative to height.  
**Insulin:** Insulin level in the blood.  
**Age:** Age of the individual in years.  
**Gender:** Reported gender of the individual.  
**Diet_Type:** General quality or type of diet followed.  
**Exercise_Frequency:** How often the individual exercises.  
**Heredity:** Whether there is a family history of diabetes.  
**Smoking:** Whether the individual smokes.  
**Alcohol:** Level of alcohol consumption.  
**Stress_Score:**  
 Numerical score representing stress level.  
**Sleep_Hours:**   Average number of hours slept.  
**Diabetes_Status:**   Indicates whether the individual has diabetes (1 = Yes, 0 = No).

**Type of Analysis**

**Descriptive Analysis**
1. To summarize health metrics such as glucose, BMI, insulin, and age across individuals.
2. To compare diabetic vs. non-diabetic individuals based on lifestyle habits (diet type, exercise frequency, smoking, alcohol).
3. To analyze the distribution of glucose, BMI, insulin, stress score, and sleep hours across the dataset.

**Diagnostic Analysis**
1. To identify the factors that contribute most to diabetes risk (glucose, BMI, heredity, lifestyle habits).
2. To examine the relationship between glucose, BMI, insulin, stress, sleep, and diabetes status.
3. To compare risk factor patterns between individuals with and without diabetes.

### Analysis about the Diabetes Status Distribution

````python
# your chart here
# If you want more cells, you can add or less cells you can remove, but dont change the structure.
Status = df['Diabetes_Status'].value_counts().index
Count = df['Diabetes_Status'].value_counts().values

plt.pie(Count,
        labels=Status,
        colors=['skyblue','Yellow','pink'],
        autopct='%1.2f%%',
        explode=[0,0.1,0],
        shadow=True)
plt.title('Diabetes Status Distribution')
plt.show()
````


**Analysis:** Univariate Analysis

**Visualization:** Pie Chart

**Chart Explanation:**
This chart shows the distribution of patients across different Diabetes Status categories — Diabetic, Non-Diabetic, and Unknown.

**Purpose:**
To find the count and proportion of diabetic patients and understand overall diabetes prevalence within the dataset.

**Features Used:**
Diabetes_Status

**Business Insight:**
Out of all patients, 88% are diabetic, 7% are non-diabetic, and 5% have missing status.

This shows the percentage of patients in each diabetes-status category and that **the dataset is imbalanced**.


### Analysis about the Insulin Distribution

````python
sns.boxplot(
    y="Insulin",
    data=df,
    showfliers=True,
    palette="bright",
    orient="v")
plt.title("Distribution of Insulin")

plt.show()
````

**Analysis:** Univariate Analysis

**Visualization:** Boxplot

**Chart Explanation:**
This chart shows the distribution of Insulin values in the dataset, including the median, spread, and outliers.

**Purpose:**
To understand the typical range of Insulin levels and identify unusual or extreme values.

**Features Used:**
Insulin

**Business Insight:**
Insulin levels mostly fall between 65–95 (median ~80), which is normal for post-meal readings. The high and low outliers (below 5– Above 155) are still medically possible, so they should be kept rather than deleted.

## Analysis about Glucose Vs Insulin

````python
sns.scatterplot(
    data=df,
    x="Glucose",
    y="Insulin",
    hue="Diabetes_Status")
plt.title("Relationship Between Glucose and Insulin")
plt.show()
````

**Analysis:** Multivariate Analysis

**Chart Explanation:**
This chart shows the relationship between Insulin and Glucose, colored by Diabetes Status (Diabetic, Non-Diabetic, Unknown).

**Business Insight:**

### Analysis about Distribution of Alcohol by Diabetes Status

````python
sns.countplot(
    x="Alcohol",
    hue="Diabetes_Status",
    data=df,
    palette="viridis"
)
plt.title("No of Patients by Alcohol and Diabetes Status")
plt.legend(bbox_to_anchor=(1,1))
plt.show()
````


**Analysis:** Bivariate Analysis

**Visualization:** Count Plot

**Chart Explanation:** This chart shows the distribution of patients by alcohol consumption level and diabetic status.

**Purpose:** To understand how diabetic and non-diabetic patients are distributed across different alcohol consumption categories.

**Features Used:** Alcohol, Diabetes Status

**Business Insight:** The chart shows that the Unknown alcohol category has the highest number of patients, followed by Low and Moderate categories.

 Diabetic patients are present across all alcohol categories, but the distribution is much higher in the Unknown category.

 This shows an association between alcohol-consumption category and diabetes status in this dataset, but it does not mean alcohol directly causes diabetes.

````python
sns.lineplot(x="Age_Group",
    y="Glucose",
    hue="Diabetes_Status",
    errorbar=None,
    estimator='mean',
    data=df
)
plt.title("Average Glucose Level By Age Group & Diabetic Status")
plt.legend(bbox_to_anchor=(1,1))
plt.show()
````

## Interpretation of your above chart

**Visualization:** Lineplot

**Chart Explanation:**

**Purpose:**
To understand the Glucose and Age affect Diabetes

**Features Used:**
GLucose, Age_Group, Diabetes_Status.

**Business Insight:**
**Glucose is strongly related to Diabetic people** because they have higher average glucose levels, and also **Age groups do not cause a major change in the average glucose  as a dataset finding.**

````python
corr = df[['Glucose', 'BMI', 'Insulin', 'Age', 'Stress_Score', 'Sleep_Hours']].corr()

sns.heatmap(corr, annot=True)
plt.title('Correlation Heatmap of Diabetes Factors')
plt.show()
````

## Interpretation of your above chart
**Analysis:** Multivariate Analysis

**Visualization:** Heat map

**Chart Explanation:**
 It shows the correlation between Glucose, Insulin, BMI, Age, Stress_Store, Sleep hours.

**Purpose:**
To understand the relation between numerical columns.

**Features Used:**
Glucose, Insulin, BMI, Age, Stress_Store, Sleep hours

**Business Insight:**
 From this chart, **Glucose and insulin have the strongest relationship** among the numerical variables, while BMI has only a weak relationship, and age, stress, and sleep hours have no relationship with others.
